 ### Model Integration with OpenAi,Google Gemini and GROQ

--> Google Gemini Model integration

```python
import os 
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("google_genai:gemini-3.5-flash-lite")
## Invoke the model 
response=model.invoke("Hello How are you?")
response.text
```

--> OpenAi Model Integration

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("gpt-4.1")
response=model.invoke("Hello how are you?")
print(response.content)
```

### USING CHAT MODELS

 ### Alternate option :
  using ChatOpenAI (IN PLACE OF init_chat_model)


```python
from langchain_openai import ChatOpenAI
model=ChatOpenAI(model="gpt-4.1")
response=model.invoke("Hello how are you?")
print(response.content)
```

Alternate option : using ChatGoogleGenerativeAI (IN PLACE OF init_chat_model)

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model=ChatGoogleGenerativeAI(model="gemini-3.5-flash")
response=model.invoke("What do you mean by Genai?")
response.text
```

GROQ Model Integration

```python
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("How are you?")
response
```
